# Getting Started with CrowdingModeWorkshop

Welcome to the `CrowdingModeWorkshop` tutorials! The `crowding` package implements the Flory-Huggins based thermodynamic model for protein stability in crowded environments.

## What this package does
The package allows you to:
1. Simulate the effect of cosolutes (crowders) on protein stability ($\Delta\Delta G^0$, $\Delta\Delta H^0$, $T\Delta\Delta S^0$).
2. Fit experimental stability data to extract cosoute-protein soft interaction parameters ($\varepsilon$, $\varepsilon_{TS}$).

## Scientific background
The physical framework relies on the Flory-Huggins model that accounts for:
- **Excluded Volume ($\nu$):** Hard-core repulsions that always stabilize the compact/folded state.
- **Flory-Huggins Interactions ($\chi$):** Nonideal solvent-cosolute interactions.
- **Soft Interactions ($\varepsilon$):** Attractive/repulsive interactions between the protein and the cosolute.

## Available Workflows

The package uses a modular, object-oriented approach. You first define the physical properties of your system (the Protein and the Cosolutes), and then pass those objects into the core thermodynamic model to simulate stability or fit experimental data.

![Workflow Flowchart](assets/workflow.svg)

### Core Classes
- `cr.Protein`: Defines the protein's Solvent Accessible Surface Area (SASA).
- `cr.Cosolute`: Define the properties of the crowders ($\nu$ and $\chi$).
- `cr.CrowdingModel`: Simulation and fitting of crowding thermodynamics for the protein in presence of the cosolute.
- `cr.Plotter`: Provides preconfigured visualization tools for plotting crowding thermodynamics and model-fitting results.

## Units and Notation
Consistent units are critical for correct thermodynamic modeling.
- **Concentration:** The core model operates using **volume fraction ($\phi$)**. Supply concentrations in $\phi$, or if you supply molality or molarity ensure you specify the units via the concentration_type flag.
- **Potenials:** The package expects experimental potentials ($\Delta\Delta G^0$, $\Delta\Delta H^0$, or $T\Delta\Delta S^0$) to be in **kJ/mol**. If your data is in kcal/mol, convert it by multiplying by $4.184$.
- **Symbols used in the code:**
  - `dG/dA` -> $\Delta\Delta G^0$ - the change in folding / association free energy
  - `dH/dE` -> $\Delta\Delta H^0$ - the change in folding / association enthalpy
  - `dS` -> $\Delta\Delta S^0$ - the change in folding / association entropy 
  - `nu` -> $\nu$
  - `chi`, `chiH`, `chiTS` -> $\chi$, $\chi_H$, $\chi_{TS}$
  - `eps`, `epsH`, `epsTS` -> $\varepsilon$, $\varepsilon_H$, $\varepsilon_{TS}$
